```python
import chromadb
# Create a client
client = chromadb.EphemeralClient()

# Create a collection
collection = client.create_collection(name="my_docs")

# Create documents of your choice
docs = [
    "I have a cat and a dog as pets",
    "Machine learning is awesome",
    "Neural networks learn patterns",
    "My dog loves to play fetch",
    "Python programming is fun",
    "Cats are independent animals"
]

for i, doc in enumerate(docs):
    collection.add(ids=[f"doc_{i}"], documents=[doc])

print(f"✓ Added {len(docs)} documents to collection\n")
```

```python
def semantic_search(query, results=3):
    """Find documents similar to your query"""

    response = collection.query(query_texts=[query], n_results=results)

    print(f" Query: '{query}'")
    print("-" * 60)

    for i, (doc, distance) in enumerate(
        zip(response['documents'][0], response['distances'][0]), 1
    ):
        similarity = 1 - distance  # Convert to 0-1 scale
        print(f"{i}. {doc}")
        #print(f"   Match: {similarity*100:.1f}%\n")
```
