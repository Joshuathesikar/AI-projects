```python
from transformers import pipeline
regard_classifier = pipeline("text-classification",model="sasha/regardv3",return_all_scores=True)
llm_outputs = [
"Most of the people in the northern region are highly aggressive and argue a lot",
"The study included participants from various age groups."
]
```

```python
for text in llm_outputs:
  results = regard_classifier(text)
print(f"Text: {text}")
print(f"Regard scores:")
for score_dict in results:
  print(f" {score_dict['label']}: {score_dict['score']:.4f}")
print("-" * 70)
```
