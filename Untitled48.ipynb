```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import os

# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Configuration & Hyperparameters
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
BATCH_SIZE = 16
LEARNING_RATE = 0.001
EPOCHS = 5

# 3. Data Preprocessing
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# --- ACTION REQUIRED: Replace with your Drive folder path ---
# Example: '/content/drive/MyDrive/MedicalData/train'
data_path = '/content/drive/MyDrive/dataaaaaaa/chest_xray/train'

if os.path.exists(data_path):
    try:
        train_dataset = torchvision.datasets.ImageFolder(root=data_path, transform=transform)
        train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
        print(f"✅ Success! Drive folder found: {data_path}")
        print(f"✅ Loaded {len(train_dataset)} images.")

        # Display sample
        img, label = train_dataset[0]
        plt.imshow(img.permute(1, 2, 0))
        plt.title(f"Sample Diagnosis: {train_dataset.classes[label]}")
        plt.show()
    except Exception as e:
        print(f"❌ Error: {e}")
else:
    print(f"❌ Error: The path {data_path} does not exist. Check your Google Drive folder name.")

# 4. Defining the "Medical" CNN Architecture
class DiagnosticCNN(nn.Module):
    def __init__(self):
        super(DiagnosticCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 32 * 32, 128)
        self.fc2 = nn.Linear(128, 2)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(-1, 32 * 32 * 32)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# 5. Initialize Model
model = DiagnosticCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)

print(f"Model initialized on: {device}")
```

```python
import time

print(f"Starting Training on {device}...")
model.train() # Set model to training mode

for epoch in range(EPOCHS):
    running_loss = 0.0
    start_time = time.time()

    for images, labels in train_loader:
        # Move data to GPU if available
        images, labels = images.to(device), labels.to(device)

        # 1. Clear previous gradients
        optimizer.zero_grad()

        # 2. Forward pass: Make a prediction
        outputs = model(images)

        # 3. Calculate loss (how wrong was the guess?)
        loss = criterion(outputs, labels)

        # 4. Backward pass: Calculate how to improve
        loss.backward()

        # 5. Optimize: Update the model weights
        optimizer.step()

        running_loss += loss.item()

    # Print statistics for the epoch
    avg_loss = running_loss / len(train_loader)
    duration = time.time() - start_time
    print(f"Epoch [{epoch+1}/{EPOCHS}] - Loss: {avg_loss:.4f} - Time: {duration:.2f}s")

print("Finished Training!")
```

```python
import torch.nn.functional as F

# 1. Set model to evaluation mode
model.eval()

# 2. Get a single batch of images from your data
images, labels = next(iter(train_loader))
images, labels = images.to(device), labels.to(device)

# 3. Make a prediction
with torch.no_grad():
    outputs = model(images)
    # Convert output scores to probabilities
    probabilities = F.softmax(outputs, dim=1)
    # Find the class with the highest probability
    _, predicted = torch.max(outputs, 1)

# 4. Visualize the Result
plt.figure(figsize=(10, 5))
for i in range(4): # Show first 4 images from the batch
    plt.subplot(1, 4, i+1)
    # Un-normalize image for viewing
    img = images[i].cpu().permute(1, 2, 0) * 0.5 + 0.5
    plt.imshow(img)

    actual_label = train_dataset.classes[labels[i]]
    pred_label = train_dataset.classes[predicted[i]]
    confidence = probabilities[i][predicted[i]].item() * 100

    title_color = 'green' if actual_label == pred_label else 'red'
    plt.title(f"Act: {actual_label}\nPred: {pred_label}\n({confidence:.1f}%)", color=title_color)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
